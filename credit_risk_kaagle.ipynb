Descrição de Variáveis

person_age: Idade da pessoa que está solicitando o empréstimo.

person_income: Renda anual da pessoa.

person_home_ownership: Tipo de propriedade ou condição de moradia da pessoa.
- rent: A pessoa mora em um imóvel alugado.
- mortgage: A pessoa possui um imóvel financiado, com uma hipoteca ou dívida associada.
- own: A pessoa possui imóvel próprio, totalmente quitado.
- other: Outras categorias de moradia que podem ser específicas do conjunto de dados.

person_emp_length: Tempo de experiência profissional da pessoa, em anos.

loan_intent: Finalidade do empréstimo solicitado.

loan_grade: Classificação atribuída ao empréstimo com base no risco de crédito do solicitante.
- A: A pessoa apresenta excelente perfil de crédito, indicando baixo risco.
- B: A pessoa apresenta risco relativamente baixo, mas possui um perfil de crédito inferior ao da categoria A.
- C: A pessoa apresenta um perfil de crédito moderado.
- D: A pessoa apresenta um risco maior em comparação com as categorias anteriores.
- E: A pessoa apresenta um perfil de crédito mais fraco, indicando maior risco.
- F: A pessoa apresenta um risco de crédito significativo.
- G: A pessoa apresenta o pior perfil de crédito entre as categorias, indicando o maior risco.

loan_amnt: Valor do empréstimo solicitado pela pessoa.

loan_int_rate: Taxa de juros associada ao empréstimo.

loan_status: Situação do empréstimo, em que 0 indica ausência de inadimplência e 1 indica inadimplência.
- 0 — Não inadimplente: A pessoa pagou o empréstimo conforme os termos acordados, sem entrar em inadimplência.
- 1 — Inadimplente: A pessoa não cumpriu as condições de pagamento acordadas e entrou em inadimplência.

loan_percent_income: Percentual da renda anual da pessoa que é representado pelo valor do empréstimo.

cb_person_default_on_file: Histórico de inadimplência da pessoa, conforme os registros dos órgãos de proteção ao crédito.
- Y (Yes): A pessoa possui histórico de inadimplência registrado em seu histórico de crédito.
- N (No): A pessoa não possui histórico de inadimplência registrado.

cb_preson_cred_hist_length: Tempo de histórico de crédito da pessoa, em anos.
